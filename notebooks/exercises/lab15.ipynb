## 7. 상자그림과 관측점을 그리는 함수

```python
def box_with_points(ax, data, group_col, value_col, order, seed=0):
    groups = [data.loc[data[group_col] == g, value_col].values for g in order]
    ax.boxplot(groups, tick_labels=order, widths=____, showfliers=False)
    rng = np.random.default_rng(seed)
    for i, values in enumerate(groups, start=1):
        jitter = rng.normal(i, 0.05, len(values))
        ax.scatter(jitter, values, s=14, alpha=0.4, color="black")
    return ax
```

## 8. 일반과 스테비아의 분포 비교하기

```python
fig, ax = plt.subplots(figsize=(5, 3))
box_with_points(ax, tomato, 
                "cultivation", "brix", 
                ____)
ax.set(xlabel="재배방식", ylabel="당도 (Brix)",
       title="일반 재배와 스테비아 재배의 당도")
plt.show()
```
