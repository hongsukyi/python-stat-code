## 3. 상자그림과 관측점을 그리는 함수

```python
def box_with_points(ax, data, group_col, value_col, order, seed=0):
    groups = [data.loc[data[group_col] == g, value_col].values for g in order]  
    ax.boxplot(groups, tick_labels=order, widths=____, showfliers=False)

    rng = np.random.default_rng(seed)
    for i, values in enumerate(groups, start=1):
        jitter = rng.normal(i, 0.05, len(values))
        ax.scatter(jitter, values, s=14, alpha=0.4, color="black")

    return ax
```

## 4. 세 집단의 당도 분포 비교하기

```python
order = ["일반", "스테비아", "프리미엄"]
fig, ax = plt.subplots(figsize=(5, 3))
box_with_points(ax, tomato, "cultivation", ____, order)
ax.set(xlabel="재배방식", ylabel="당도 (Brix)",
       title="재배방식별 당도 분포")
plt.show()
```

## 5. 집단 내 흩어짐 비교하기

```python
rng = np.random.default_rng(____)

fig, axes = plt.subplots(1, 2, figsize=(7, 3), sharey=True)

for ax, spread, title in [(axes[0], 0.4, "집단 내 흩어짐이 작다"),
                          (axes[1], 1.8, "집단 내 흩어짐이 크다")]:
    for i, center in enumerate([10, 11, 12], start=1):
        values = rng.normal(center, spread, 30)
        jitter = rng.normal(i, 0.06, 30)
        ax.scatter(jitter, values, s=16, alpha=0.7)

    ax.set(title=title, xticks=[1, 2, 3], xticklabels=["A", "B", "C"])

axes[0].set_ylabel("값")
plt.tight_layout()
plt.show()
```

## 6. 일원분산분석 실행하기

```python
groups = { name: tomato.loc[tomato["cultivation"] == name, ____].values            
    for name in order}
f_stat, p_value = stats.f_oneway(*(groups[name] for name in order))
print("F 통계량:", round(f_stat, 3))
print("p값     :", format(p_value, ".2e"))
```

## 7. Tukey 사후검정

```python
if p_value < ____:
    tukey = stats.tukey_hsd(*(groups[name] for name in order))
    rows = []
    for i in range(len(order)):
        for j in range(i + 1, len(order)):
            rows.append({
                "집단 1": order[i], "집단 2": order[j],
                "평균 차이": groups[order[j]].mean() - groups[order[i]].mean(),
                "조정 p값": tukey.pvalue[i, j], "차이 있음": tukey.pvalue[i, j] < 0.05, }) 
    tukey_table = pd.DataFrame(rows).round(4)
    print(tukey_table.to_string(index=False))
else:
    print("전체 차이가 확인되지 않아 사후검정을 생략합니다.")
```
