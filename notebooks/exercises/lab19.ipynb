## 5. 세 그래프로 차이와 관계 살펴보기

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4.2))
order = ["일반", "스테비아", "프리미엄"]

brix_groups = [tomato.loc[tomato["cultivation"] == g, "brix"].values for g in order] 
axes[0].boxplot(brix_groups, tick_labels=order, widths=0.5)
axes[0].set(title="재배방식별 당도", xlabel="재배방식", ylabel="당도 (Brix)")

sales_table = pd.crosstab(tomato["color"], tomato["status"])
sales_table = sales_table.loc[["빨강", "주황", "노랑"]]
x = np.arange(len(sales_table))
width = ____
axes[1].bar(x - width / 2, sales_table["완판"], width, label="완판")
axes[1].bar(x + width / 2, sales_table["재고"], width, label="재고")
axes[1].set(title="색깔별 판매", xlabel="색깔", ylabel="개수", xticks=x, 
            xticklabels=sales_table.index)
axes[1].legend(title="판매결과")

axes[2].scatter(tomato["sunlight"], tomato["brix"], alpha=0.65)
axes[2].set(title="일조시간과 당도", xlabel="일조시간", ylabel="당도 (Brix)")
plt.tight_layout()
plt.show()
```

## 6. 재배방식별 당도 ANOVA

```python
groups = [tomato.loc[tomato["cultivation"] == name, ____] for name in order]
anova = stats.f_oneway(*groups)
print(f"ANOVA: F={anova.statistic:.3f}, p={anova.pvalue:.4g}")
```

## 7. Tukey 사후검정

```python
if anova.pvalue < ____:
    tukey = stats.tukey_hsd(*groups)
    for i, j in [(0, 1), (0, 2), (1, 2)]:
        print(f"{order[i]}-{order[j]}: p={tukey.pvalue[i, j]:.3f}")
else:
    print("전체 차이가 확인되지 않아 사후검정을 생략합니다.")
```

## 8. 색깔과 판매결과 분석하기

```python
color_table = pd.crosstab(tomato[____], tomato["status"])
color_test = stats.chi2_contingency(color_table)
color_rate = pd.crosstab(tomato["color"], tomato["status"], normalize="index")["완판"] * 100

print(color_rate.round(1).astype(str).add("%").to_string())
print(f"카이제곱={color_test.statistic:.3f}, p={color_test.pvalue:.4f}")
```

## 9. 재배방식과 판매결과 분석하기

```python
cultivation_table = pd.crosstab(tomato[____], tomato["status"])
cultivation_test = stats.chi2_contingency(cultivation_table)
cultivation_rate = pd.crosstab(tomato["cultivation"], tomato["status"], normalize="index")["완판"] * 100

print(cultivation_rate.round(1).astype(str).add("%").to_string())
print(f"카이제곱={cultivation_test.statistic:.3f}, p={cultivation_test.pvalue:.4f}")
```

## 10. 상관계수와 회귀식 구하기

```python
regression = stats.linregress(tomato[____], tomato["brix"])
print(f"r={regression.rvalue:.3f}")
print(f"예측 당도={regression.intercept:.3f}+{regression.slope:.3f}×일조시간")
print(f"R²={regression.rvalue ** 2:.3f}")
```

## 11. 분석 결과 종합표 만들기

```python
result_summary = pd.DataFrame({
    "분석": ["재배방식-당도", "색깔-판매결과", "재배방식-판매결과", "일조시간-당도"],
    "통계 결과": [
        f"F={anova.statistic:.2f}, p={anova.pvalue:.3g}",
        f"χ²={color_test.statistic:.2f}, p={color_test.pvalue:.3f}",
        f"χ²={cultivation_test.statistic:.2f}, p={cultivation_test.pvalue:.3f}",
        f"r={regression.rvalue:.3f}, R²={regression.rvalue ** 2:.3f}"
    ],
    "판단": [
        "평균 차이 확인" if anova.pvalue < 0.05 else "평균 차이 확인 못함",
        "관련성 확인" if color_test.pvalue < 0.05 else "관련성 확인 못함",
        "관련성 확인" if cultivation_test.pvalue < 0.05 else "관련성 확인 못함",
        "양의 직선 관계" if regression.rvalue > 0 else "음의 직선 관계" if regression.rvalue < 0 else "직선 관계 없음"
    ]
})
print(result_summary.to_string(index=____))
```
