## 6. 숫자 집계하기

```python
scores = {"홍길동": ____, "이순신": 92, "강감찬": 79}

print("합계:", sum(scores.values()))
print("최고:", max(scores.values()))
print("평균:", round(sum(scores.values()) / len(scores), 2))
```
