## 9. 산점도에 직선 추세 추가하기

```python
slope, intercept = np.polyfit(df["sunlight"], 
                              df["brix"], 1)
x_line = np.linspace(df["sunlight"].min(), 
                     df["sunlight"].max(), 100) 
y_line = intercept + slope * x_line

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(df["sunlight"], df["brix"], 
           alpha=0.5, label="관측값")
ax.plot(x_line, y_line, 
        linewidth=____, label="직선 추세")
ax.set(title="일조시간과 당도의 직선 추세", 
       xlabel="하루 평균 일조시간", 
       ylabel="당도 (Brix)")
ax.legend();plt.show()
print("기울기:", round(slope, 3))
```
