## 5. 단순선형회귀식 구하기

```python
regression = stats.linregress(tomato[____], 
                              tomato["brix"])  
slope = regression.slope
intercept = regression.intercept

print("기울기:", round(slope, 3))
print("절편  :", round(intercept, 3))
print("예측={:.3f}+{:.3f}×시간".format(intercept, 
                                   slope))
```

## 7. 산점도에 회귀선 추가하기

```python
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(tomato["sunlight"], tomato["brix"], 
           alpha=0.6, s=45) 
line_x = np.linspace(tomato["sunlight"].min(), 
                     tomato["sunlight"].max(),50)  
line_y = intercept + slope * line_x

ax.plot(line_x, line_y, color="tomato", 
        linewidth=____, label="회귀직선")
ax.set(
    xlabel="하루 평균 일조시간",
    ylabel="당도 (Brix)",
    title="당도 예측 회귀")
ax.set_xlim(0,15)
ax.set_ylim(0,20)
ax.legend(loc='upper left')
plt.show()
```
