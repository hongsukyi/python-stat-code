## 12. 등급별 속도 분포 비교하기

```python
order = ["원활", "혼잡"]
groups = [prepared.loc[prepared[____] == g, "Speed"] for g in order]

fig, ax = plt.subplots(figsize=(5, 3))
ax.boxplot(groups, tick_labels=order, showfliers=False)
rng = np.random.default_rng(0)
for i, values in enumerate(groups, start=1):
    ax.scatter(rng.normal(i, 0.05, len(values)), values,
               s=1, alpha=0.1, color="gray")
ax.set(title="등급별 속도 분포", xlabel="등급", ylabel="속도 (km/h)")
plt.show()
```
